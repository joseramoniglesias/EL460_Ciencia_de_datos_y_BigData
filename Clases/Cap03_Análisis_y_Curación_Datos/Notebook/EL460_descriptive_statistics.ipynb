Algunos estdísticos comunes y usuales. Por ejemplo podemos calcular la media, mediana y moda, así como también las desviaciones estándar y varianzas para cada una de las 7 variables.

- Media (Promedio):
\begin{equation}
\bar x=\frac{1}{N}\sum_{i=0}^{N-1}x_i
\end{equation}
- Mediana (Cuartil 50): Con datos ordenados en orden ascendente:
\begin{equation}
median=\left\{\begin{matrix}
x\left[\frac{N}{2}\right] &  \mbox{si $N$ es par}\\
\frac{1}{2}\left(x\left[\frac{N-1}{2}\right]+x\left[\frac{N+1}{2}\right]\right) &  \mbox{si $N$ es impar}\\
\end{matrix}\right.
\end{equation}
- Moda (dato con mayor frecuencia)
- varianza:
\begin{equation}
s^2(x)=\mbox{Var}(x)=\frac{1}{N}\sum_{i=0}^{N-1}(x_i-\bar x)^2
\end{equation}
- Desviación estándar:
\begin{equation}
\sigma_x=s(x)=\sqrt{\mbox{Var}(x)}=\sqrt{\frac{1}{N}\sum_{i=0}^{N-1}(x_i-\bar x)^2}
\end{equation}


```python
def average(var):
    return np.sum(var)/len(var)
def median(var):
    var_ord=np.sort(var)
    N=len(var_ord)
    if N%2==0:
        median=var_ord[N//2]
    if N%2!=0:
        median=(var_ord[(N-1)//2]+var_ord[(N+1)//2])/2
    return median
def mode(var):
    dat,frec=np.unique(var,return_counts=True)
    if len(dat[frec==frec.max()])>1:
        print("Existen varias modas (la variable es multimodal)")
    if len(dat[frec==frec.max()])==1:
        print("Existe una moda (la variable es unimodal)")
    return dat[frec==frec.max()]
```

```python
def variance(var):
    return np.sum((var-average(var))**2)/len(var)
def std(var):
    return np.sqrt(variance(var))
```

```python
ionosphere_stats=data.describe().T
```

```python
ionosphere_stats["LII"]=ionosphere_stats["25%"]-((ionosphere_stats["75%"]-ionosphere_stats["25%"])*1.5)
ionosphere_stats["LIS"]=ionosphere_stats["75%"]+((ionosphere_stats["75%"]-ionosphere_stats["25%"])*1.5)
ionosphere_stats["LEI"]=ionosphere_stats["25%"]-((ionosphere_stats["75%"]-ionosphere_stats["25%"])*3.0)
ionosphere_stats["LES"]=ionosphere_stats["75%"]+((ionosphere_stats["75%"]-ionosphere_stats["25%"])*3.0)
```

```python
ionosphere_stats
```

```python
# 1. Inicializar el DataFrame copiando el índice directamente
ionosphere_stats2 = pd.DataFrame(index=ionosphere_stats.index)

atipicos, extremos = [], []
total_filas = len(data)

# 2. Iterar eficientemente usando .itertuples() o .loc
for i in ionosphere_stats2.index:
    # Extraer los límites de forma directa y rápida
    LII = ionosphere_stats.loc[i, "LII"]
    LIS = ionosphere_stats.loc[i, "LIS"]
    LEI = ionosphere_stats.loc[i, "LEI"]
    LES = ionosphere_stats.loc[i, "LES"]

    columna = data[i]

    # CORRECCIÓN DE LÓGICA:
    # Atípicos: Entre el límite externo inferior y el interno inferior,
    # O entre el límite interno superior y el externo superior.
    condicion_atipicos = ((columna >= LEI) & (columna < LII)) | ((columna > LIS) & (columna <= LES))
    atipicos.append(condicion_atipicos.sum())

    # Extremos: Estrictamente fuera de los límites externos
    condicion_extremos = (columna < LEI) | (columna > LES)
    extremos.append(condicion_extremos.sum())

# 3. Asignar columnas calculadas
ionosphere_stats2["Valores atipicos"] = atipicos
ionosphere_stats2["Valores extremos"] = extremos
ionosphere_stats2["Valores atipicos (%)"] = np.round((ionosphere_stats2["Valores atipicos"] / total_filas) * 100, 2)
ionosphere_stats2["Valores extremos (%)"] = np.round((ionosphere_stats2["Valores extremos"] / total_filas) * 100, 2)

# 4. Agregar fila de totales (calculando correctamente los porcentajes del gran total)
totales = ionosphere_stats2[['Valores atipicos', 'Valores extremos']].sum()
ionosphere_stats2.loc['Total'] = [
    totales['Valores atipicos'],
    totales['Valores extremos'],
    np.round((totales['Valores atipicos'] / total_filas) * 100, 2),
    np.round((totales['Valores extremos'] / total_filas) * 100, 2)
]
```

Tomemos por ejemplo los datos que tienen valores atípicos en mayor cantidad, por ejemplo, los datos con  `Atr33`, y veamos como cambia la distribución cual eliminamos los valores atípicos.

```python
def filter_outlier(col):
    LII = ionosphere_stats.loc[col, "LII"]
    LIS = ionosphere_stats.loc[col, "LIS"]
    LEI = ionosphere_stats.loc[col, "LEI"]
    LES = ionosphere_stats.loc[col, "LES"]
    return data[~(((data[col]<=LII)&(data[col]>=LIS))|((data[col]>=LIS)&(data[col]<=LES)))][col]
```

```python
# 1. Almacenar variables para optimizar rendimiento y consistencia
data_all = data["Atr33"]
data_filtered = filter_outlier("Atr33")

len_all, len_filt = len(data_all), len(data_filtered)
mean_all, mean_filt = np.mean(data_all), np.mean(data_filtered)
median_all, median_filt = np.median(data_all), np.median(data_filtered)

plt.figure(figsize=(16, 6))

# --- SUBPLOT 1: Distribución Absoluta ---
plt.subplot(121)
# Se invierte el orden (primero total, luego filtrado) para mejorar la mezcla visual
counts_all, bins_all, _ = plt.hist(data_all, bins=20, label="Atr33 (All data) ({} rows)".format(len_all), alpha=0.4, color='r')
counts_filt, bins_filt, _ = plt.hist(data_filtered, bins=20, label="Atr33 (Outliers removed) ({} rows)".format(len_filt), alpha=0.4, color='b')

# Altura dinámica para las líneas del primer gráfico (un 5% más alto que el bin máximo)
ymax_abs = max(counts_all.max(), counts_filt.max()) * 1.05

plt.vlines(mean_filt, 0, ymax_abs, label="Outliers removed mean (<Atr33>={})".format(np.round(mean_filt, 4)), alpha=0.99, color='b', linestyle="--", linewidth=4)
plt.vlines(mean_all, 0, ymax_abs, label="All data mean (<Atr33>={})".format(np.round(mean_all, 4)), alpha=0.99, color='r', linestyle="--", linewidth=4)

plt.legend(title="Absolute distribution", fontsize=11)
plt.ylabel("Counts", fontsize=14)
plt.xticks(fontsize=14)
plt.yticks(fontsize=14)
plt.ylim(0, ymax_abs) # Forzar el límite para que coincida con la altura calculada

# --- SUBPLOT 2: Distribución Acumulada ---
plt.subplot(122)
plt.hist(data_all, bins=20, label="Atr33 (All data) ({} rows)".format(len_all), alpha=0.4, cumulative=True, color='r')
plt.hist(data_filtered, bins=20, label="Atr33 (Outliers removed) ({} rows)".format(len_filt), alpha=0.4, cumulative=True, color='b')

# Altura dinámica para el gráfico acumulado (total de registros más un margen del 5%)
ymax_cum = len_all * 1.05

# CORRECCIÓN DE LÓGICA: Ahora calcula y muestra el valor numérico de la mediana real
plt.vlines(median_filt, 0, ymax_cum, label="Outliers removed median (<Atr33>={})".format(np.round(median_filt, 4)), alpha=0.99, color='b', linestyle="--", linewidth=4)
plt.vlines(median_all, 0, ymax_cum, label="All data median (<Atr33>={})".format(np.round(median_all, 4)), alpha=0.99, color='r', linestyle="--", linewidth=4)

plt.legend(title="Cumulative distribution", fontsize=11)
plt.ylabel("Counts", fontsize=14)
plt.xticks(fontsize=14)
plt.yticks(fontsize=14)
plt.ylim(0, ymax_cum)

# Guardar y mostrar
plt.savefig("../figures/ionosphere_hist2.png", bbox_inches="tight")
plt.show()
```
